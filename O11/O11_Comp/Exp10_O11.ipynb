## Experimental Setup

The experiment was designed to investigate the behavior of polarized light using a combination of optical elements and to characterize unknown polarization states. The full setup (as depicted in the experiment manual) consisted of the following components arranged sequentially:

1. **Collimated LED Light Source**
   The source emitted monochromatic light with a peak wavelength $ \lambda_{\text{peak}} = 590, \\text{nm} $. The emitted light was initially unpolarized.

2. **Polariser or Black Box**
   For Task 1, a **linear polariser** was placed after the LED to generate a well-defined linear polarization state.
   For Task 2, this component was replaced with one of the three **black boxes** containing unknown optical elements (e.g., linear polariser, circular polariser, etc.).

3. **λ/4 (Quarter-Wave) Plate**
   This optical element introduces a phase difference between the orthogonal polarization components of the incident light. Its fast axis was rotatable and set to specific angles $ \phi = 0^\circ$, $ 30^\circ $, and $ 45^\circ $ for different measurements in Task 1.

4. **Linear Analyser**
   A rotatable polariser (also referred to as the analyser) was placed after the λ/4 plate. The transmitted light intensity was measured as a function of its orientation angle $ \\theta $.

5. **Lens**
   This focused the transmitted light onto the detection element.

6. **Photodiode + Digital Multimeter**
   The photodiode converted light intensity into a voltage signal. A digital multimeter was used to record this signal, which is directly proportional to the transmitted light intensity.

### Task 1: Procedure for Investigating Polarised Light through a λ/4 Plate

This part of the experiment aimed to compare theoretical and experimental results for light intensity as a function of analyser angle \\( \\theta \\), after a linearly polarised beam passed through a λ/4 plate.

Steps:

1. Set up the LED, linear polariser (to vertically polarise the light), λ/4 plate, analyser, and photodiode in order.
2. Fix the polarisation direction of the incoming light (typically vertical).
3. Rotate the λ/4 plate to each of the specified angles:
   - $ \phi = 0^\circ $
   - $ \phi = 30^\circ $
   - $ \phi = 45^\circ $
4. For each λ/4 plate orientation, rotate the analyser in steps (e.g., 5° or 10° increments) through a full 180° range.
5. At each analyser angle, record the light intensity measured by the photodiode.
6. Repeat for all three angles of the λ/4 plate.
7. Plot the measured intensity versus analyser angle for each case.
8. Compare the results to theoretical predictions using either nonlinear fits or analytical expressions derived from Jones matrix formalism.

### Task 2: Identification of Polarisation States Using Black Boxes

This task focused on determining the polarization state of light after it passed through unknown "black box" elements.

For this, the **simple operational guide** from the end of the lab manual was followed:

1. Replace the polariser with one of the black boxes.
2. Record the intensity as the **analyser alone** is rotated over 180°:
   - If there is **no variation** in intensity, insert the λ/4 plate before the analyser.
   - If the intensity still shows **no variation** → the light is **unpolarised**.
   - If an **intensity minimum** appears:
     - If the intensity goes to **zero** at some angle → **circularly polarised**.
     - If not → **mixture of circular and unpolarised light**.
3. If **intensity variation** is observed with the analyser alone:
   - If a **zero** appears at some angle → **linearly polarised**.
   - If no zero appears, insert the λ/4 plate before the analyser and align its fast axis with the analyser direction at maximum or minimum intensity:
     - If **zero** intensity appears after insertion → **elliptically polarised**.
     - If not, compare minimum positions with and without λ/4 plate:
       - If unchanged → **linear + unpolarised light**
       - If shifted → **elliptical + unpolarised light**
4. Repeat the procedure for all three black boxes.
5. Use observations to classify the polarization state of each black box, as well as provide supporting measurements.

## Theoretical Background

Understanding and analyzing the polarization of light requires a mathematical formalism to describe different polarization states and their transformation through optical components. The theoretical background of this experiment is built upon Jones calculus, Stokes parameters, and the Poincaré sphere representation, all of which were directly applied in the experimental tasks.

### 1. Jones Vectors and Polarisation States

A monochromatic light wave propagating in the z-direction has its electric field vector confined to the x–y plane and can be written as:

$$
\vec{E} =
\begin{pmatrix}
E_x \\\\
E_y
\end{pmatrix}
=
\begin{pmatrix}
E_{0x} e^{-i\delta_x} \\\\
E_{0y} e^{-i\delta_y}
\end{pmatrix}
$$

Neglecting the common phase factor $ e^{-i(\omega t - kz)} $, and choosing $ \delta_x = 0 $, the normalized form (called the **Jones vector**) becomes:

$$
\vec{J} =
\frac{1}{\sqrt{E_{0x}^2 + E_{0y}^2}}
\begin{pmatrix}
E_{0x} \\\\
E_{0y} e^{-i\delta}
\end{pmatrix}
$$

Different polarization states are characterized by specific values of $ E_{0x} $, $ E_{0y} $, and $ \delta $:

- **Linear polarisation**: $ \delta = 0 $ or $ \pi $
- **Circular polarisation**: $ E_{0x} = E_{0y} $, $ \delta = \pm\pi/2 $
- **Elliptical polarisation**: General case with arbitrary $ E_{0x} $, $ E_{0y} $, and $ \delta $

### 2. Jones Matrices of Optical Components

Each optical component in the setup affects the light’s polarization state and is modeled by a Jones matrix.

** Linear Polariser at Angle θ **
Projects the electric field vector onto an axis at angle $\theta$ with respect to the vertical (y-axis):
$$ M_{\theta} = \begin{pmatrix} \sin^2 \theta & \cos \theta \sin \theta \\ \cos \theta \sin \theta & \cos^2 \theta \end{pmatrix} $$

Applied to vertically polarised light ($ \vec{J} = \begin{pmatrix} 0 \\ 1 \end{pmatrix} $), the transmitted intensity follows Malus' law:
$$ I_{\theta} \propto \cos^2 \theta $$

**Quarter-Wave Plate (λ/4 Plate)**
Introduces a phase shift of $\pi/2$ between the fast and slow axes. For a fast axis aligned along x, its matrix is:
$$ M_{\lambda/4,x} = e^{-i\pi/4} \begin{pmatrix} 1 & 0 \\ 0 & i \end{pmatrix} $$

When rotated by an angle $\phi$ from vertical, the plate's matrix becomes:
$$ M_{\lambda/4,\phi} = D(\phi) M_{\lambda/4,x} D^T(\phi) $$
where $$ D(\phi) = \begin{pmatrix} \sin \phi & -\cos \phi \\ \cos \phi & \sin \phi \end{pmatrix} $$

This formalism allows predicting how linear light becomes elliptical or circular after passing through a rotated λ/4 plate — essential for Task 1.

### 3. Stokes Parameters and Poincaré Sphere
While Jones calculus is ideal for fully polarized light, Stokes parameters are better suited for experimental characterization, especially for partially polarized or unknown states like those in Task 2.

Defined as:
$$ s_0 = E_0x^2 + E_0y^2 $$
$$ s_1 = E_0x^2 - E_0y^2 $$
$$ s_2 = 2 E_0x E_0y \cos(\delta) $$
$$ s_3 = 2 E_0x E_0y \sin(\delta) $$

These parameters form a vector $ \vec{s} = (s_0, s_1, s_2, s_3) $
which lies on the Poincaré sphere.

Key interpretations:
- **Equator**: Linear polarisation states
- **North/South poles**: Circular polarisation states
- **Points on the sphere**: Elliptical polarisation states

### 4. Practical Classification of Polarisation (Task 2 Guide)
To determine the polarisation state of an unknown beam (e.g. from a black box), the experiment used a step-by-step classification guide based on:
- Variation of intensity with analyser rotation
- Effect of inserting a λ/4 plate
- Presence or absence of minima (zero intensity)

| Observation                                          | Conclusion                     |
| ---------------------------------------------------- | ------------------------------ |
| No variation, even with λ/4 plate                    | Unpolarised light              |
| Constant with analyser; variation with λ/4 plate     | Circular or partially circular |
| Intensity zero at some analyser angle                | Linear polarisation            |
| No zero even after λ/4 plate                         | Elliptical or mixture          |
| Shift of intensity minimum after inserting λ/4 plate | Elliptical + unpolarised       |


This method allowed qualitative identification of polarization without computing full Jones or Stokes formalism, which is particularly useful for real-world beam characterizations.

## Analysis and Results

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def parabola(x,a,b,c):
    return a*(x**2)+b*x+c

data1 = pd.read_excel('O11.xlsx', sheet_name='Task1')

V_0 = data1['Voltage (V)']*1e3
V_30 = data1['Voltage (V).1']*1e3
V_45 = data1['Voltage (V).2']*1e3

D_Angle_0 = data1['Delta Angle (°)']
D_Angle_30 = data1['Delta Angle (°).1']
D_Angle_45 = data1['Delta Angle (°).2']

popt0,_ = curve_fit(parabola, D_Angle_0, V_0)
x0 = np.linspace(min(D_Angle_0),max(D_Angle_0),1000)
model0 = parabola(x0, *popt0)

popt30,_ = curve_fit(parabola, D_Angle_30, V_30)
x30 = np.linspace(min(D_Angle_30),max(D_Angle_30),1000)
model30 = parabola(x30, *popt30)

popt45,_ = curve_fit(parabola, D_Angle_45, V_45)
x45 = np.linspace(min(D_Angle_45),max(D_Angle_45),1000)
model45 = parabola(x45, *popt45)

plt.figure(1)
plt.title('Intensity per angle difference \n'r'$\theta_0$ = 0°')
plt.xlabel('Angle (°)')
plt.ylabel('Light intensity')
plt.plot(D_Angle_0,V_0, 'x', label = 'Measured intensity')
plt.plot(x0, model0, label = f'Fitted curve: y = {"%.3f"%popt0[0]}*x²+{"%.3f"%popt0[1]}*x+{"%.3f"%popt0[2]}')
plt.legend(loc = 'lower left')

plt.figure(2)
plt.title('Intensity per angle difference \n'r'$\theta_0$ = 30°')
plt.plot(D_Angle_30,V_30, 'x', label = 'Measured intensity')
plt.plot(x30, model30, label = f'Fitted curve: y = {"%.3f"%popt30[0]}*x²+{"%.3f"%popt30[1]}*x+{"%.3f"%popt30[2]}')
plt.legend(loc = 'lower left')

plt.figure(3)
plt.title('Intensity per angle difference \n'r'$\theta_0$ = 45°')
plt.plot(D_Angle_45,V_45, 'x', label = 'Measured intensity')
plt.plot(x45, model45, label = f'Fitted curve: y = {"%.3f"%popt45[0]}*x²+{"%.3f"%popt45[1]}*x+{"%.3f"%popt45[2]}')
plt.legend(loc = 'lower left')

plt.show()
```
