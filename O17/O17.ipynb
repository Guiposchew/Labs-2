## Setup
In this experiment, we investigate Fraunhofer diffraction patterns resulting from various slit configurations using a laser-based optical setup. The setup includes:
- A linearly polarized laser as the light source.
- A slide holder, which can hold interchangeable thin plates with different slit configurations: single, double, triple, and quadruple slits.
- A collimating lens to ensure a plane wavefront hits the diffraction object.
- A CCD camera mounted on a micrometer-driven rail, used to record the intensity distribution of the diffraction patterns.
- The distance $d$ between the slit and the camera, and the focal length $f$ of the lens, are adjustable parameters that influence the measurement.
- The camera is connected to a computer running “Splicco” software, used to adjust integration time and averaging settings. The data is exported as a .csv file with pixel number and corresponding intensity values.

Alignment is crucial: all optical components, especially the camera, were aligned in height and along the optical axis to minimize aberrations and maximize measurement accuracy. Data is typically collected in a darkened environment to reduce ambient light noise.

## Theoretical Background
The experiment explores Fraunhofer diffraction, which occurs when a parallel beam of light (e.g., laser light) passes through slits and the resulting far-field diffraction pattern is analyzed.

### 1. Fourier Transform Relation
Fraunhofer diffraction patterns correspond to the Fourier transform of the aperture function:
- For a single slit of width bb, the intensity distribution I(θ)I(θ) is given by:
    $$ I(\theta) \propto \left( \frac{\sin \beta}{\beta} \right)^2 ; \beta = \frac{b \pi}{\lambda} \sin \theta $$
- For multiple slits, interference terms modulate this envelope, introducing additional minima and maxima depending on slit spacing $g$ and number of slits.

The spatial pattern on the CCD camera corresponds to $I(x)$, where $x$ is related to $\theta$ through the geometry of the setup:
$$ x = f \tan(\theta) \approx f \theta \quad \text{(for small angles)} $$

### 2. Fresnel vs. Fraunhofer Regime
The diffraction observed depends on the Fresnel number $F$, defined as:
$$ F = \frac{a^2}{\lambda d} $$
where $a$ is the aperture size, $\lambda$ is the wavelength, and $d$ is the distance between the aperture and the observation screen.
- **Franhofer regime**: $F \ll 1$ (far-field diffraction).
- **Fresnel regime**: $F \gtrsim 1$ (near-field diffraction), where the curvature of the wavefronts must be considered.

In this experiment, both regimes are explored by adjusting the slit-camera distance $d$.

### 3. Fourier Analysis in Experiment
By recording the diffraction pattern and performing a Fast Fourier Transform (FFT) on the intensity data:
- The aperture width $b$ can be extracted from the frequency of minima.
- For double slits, both slit width $b$ and slit separation $g$ can be determined.
